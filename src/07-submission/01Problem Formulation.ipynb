```python

import sys
import subprocess
from pathlib import Path

try:
    from reportlab.lib import colors
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.enums import TA_JUSTIFY, TA_CENTER, TA_LEFT
    from reportlab.lib.units import inch
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "reportlab"])
    from reportlab.lib import colors
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.enums import TA_JUSTIFY, TA_CENTER, TA_LEFT
    from reportlab.lib.units import inch
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle

file_name = "01 Problem Formulation- DM4ML-Group51.pdf"
output_path = Path.cwd() / file_name

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    name="TitleStyle",
    parent=styles["Title"],
    alignment=TA_CENTER,
    fontSize=16,
    leading=20,
    spaceAfter=16,
)

heading_style = ParagraphStyle(
    name="HeadingStyle",
    parent=styles["Heading2"],
    fontSize=12,
    leading=15,
    spaceAfter=8,
    textColor=colors.black,
)

body_style = ParagraphStyle(
    name="BodyStyle",
    parent=styles["BodyText"],
    alignment=TA_JUSTIFY,
    fontSize=10.5,
    leading=16,
    spaceAfter=10,
)

meta_style = ParagraphStyle(
    name="MetaStyle",
    parent=styles["BodyText"],
    alignment=TA_LEFT,
    fontSize=10.5,
    leading=14,
    spaceAfter=6,
)

bullet_style = ParagraphStyle(
    name="BulletStyle",
    parent=styles["BodyText"],
    alignment=TA_LEFT,
    fontSize=10.3,
    leading=15,
    leftIndent=14,
    firstLineIndent=-8,
    spaceAfter=4,
)

title = "01 Problem Formulation"
course_name = "Data Management for Machine Learning"
assignment_title = "End-to-End Data Management Pipeline for a Recommendation System"
assignment_group = "Group 51 - Data management for Machine Learning Group 51"

team_data = [
    ["Team Member Name", "Team Member ID"],
    ["BANSHIDHAR RATH", "2025AE05346"],
    ["JITENDRA KUMAR TIWARI", "2025AE05518"],
    ["KATBA ANKIT CHIMANBHAI", "2025AE05229"],
    ["NAVEEN SURATHU", "2025AE05492"],
]

team_table = Table(team_data, colWidths=[3.8 * inch, 2.0 * inch])
team_table.setStyle(
    TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9EAD3")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.black),
        ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
        ("FONTNAME", (0, 1), (-1, -1), "Helvetica"),
        ("FONTSIZE", (0, 0), (-1, -1), 10),
        ("GRID", (0, 0), (-1, -1), 0.6, colors.grey),
        ("ALIGN", (0, 0), (-1, -1), "LEFT"),
        ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
        ("TOPPADDING", (0, 0), (-1, -1), 6),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 6),
    ])
)

problem_definition = """
RecoMart is an e-commerce platform that aims to improve customer engagement, cross-selling opportunities, 
and overall conversion rate through personalized product recommendations. The core business problem is to build 
a robust and scalable data management pipeline that can continuously ingest, validate, prepare, and transform 
data from multiple operational sources so that recommendation models always learn from fresh, trustworthy, and 
well-structured information. The pipeline must support the end-to-end lifecycle of recommendation analytics, 
from raw data acquisition to model-ready features and deployable model outputs. In practical terms, the system 
is expected to reduce data inconsistency, improve the quality of downstream recommendations, and enable a 
repeatable workflow for both training and inference use cases.
"""

objectives = """
The primary objective of this project is to design and implement an automated, modular, and reproducible data 
pipeline for a recommendation system. The pipeline is expected to support multi-source ingestion, structured raw 
data storage, data quality validation, remediation of identified issues, preparation of analysis-ready datasets, 
feature engineering for recommendation algorithms, and orchestration of the complete workflow. A secondary 
objective is to ensure traceability and operational reliability through logging, validation reports, and 
version-aware outputs. This is especially important in recommendation use cases, where inaccurate or incomplete 
interaction and catalog data can directly degrade model quality and business relevance.
"""

data_sources_intro = """
The recommendation pipeline integrates multiple data sources representing user behavior and product metadata. 
These sources collectively support collaborative filtering, content-based filtering, and hybrid recommendation 
use cases.
"""

data_source_bullets = [
    "<b>RetailRocket interaction data (CSV)</b> - Includes files such as <i>events.csv</i>, "
    "<i>category_tree.csv</i>, <i>item_properties_part1.csv</i>, and <i>item_properties_part2.csv</i>. "
    "These files capture user-item interaction history and item attribute information.",

    "<b>Key attributes from interaction and catalog data</b> - User-related fields such as "
    "<i>visitorid</i>; item-related fields such as <i>itemid</i>; behavioral attributes such as "
    "<i>event</i> and <i>timestamp</i>; transaction-related fields such as <i>transactionid</i>; "
    "category attributes such as <i>categoryid</i> and <i>parentid</i>; and item property fields such as "
    "<i>property</i> and <i>value</i>.",

    "<b>DummyJSON product and category data (JSON / Parquet)</b> - Includes raw API-style files such as "
    "<i>products_raw.json</i> and <i>categories_raw.json</i>, and processed bronze outputs such as "
    "<i>products.parquet</i> and <i>categories.parquet</i>.",

    "<b>Key product attributes</b> - Product identifiers and descriptive fields such as <i>id</i>, "
    "<i>title</i>, and <i>category</i>; commercial and quality-related fields such as <i>price</i>, "
    "<i>rating</i>, and <i>stock</i>.",

    "<b>Analytical role of these data sources</b> - User interaction records help capture implicit feedback "
    "signals, while product and category metadata provide the descriptive context needed for item similarity, "
    "catalog enrichment, and content-based recommendation logic."
]

expected_outputs_intro = """
The pipeline is expected to produce structured outputs that directly support analysis, feature generation, 
model development, and operational deployment.
"""

expected_output_bullets = [
    "<b>Clean datasets for Exploratory Data Analysis (EDA)</b> - The pipeline should generate validated and "
    "cleaned datasets that can be used to study interaction distributions, item popularity, category coverage, "
    "data sparsity, missing value patterns, and other exploratory insights required before model development.",

    "<b>Engineered features for collaborative and content-based models</b> - The pipeline should create reusable "
    "features such as user activity frequency, item popularity, recency of interactions, category-based item "
    "descriptors, average rating signals, and other transformed attributes that improve training quality for "
    "recommendation algorithms.",

    "<b>Deployable recommendation model and inference interface</b> - The final pipeline should support at least "
    "one recommendation model and produce artifacts that can be used during inference, such as trained model files, "
    "feature retrieval logic, candidate ranking outputs, or an interface that returns recommended products for a user.",

    "<b>Quality and governance artifacts</b> - In addition to modeling outputs, the pipeline should produce logs, "
    "validation summaries, issue reports, fix logs, and version-aware outputs to ensure reproducibility, auditability, "
    "and easier operational monitoring."
]

evaluation_metrics_intro = """
Recommendation quality must be evaluated using ranking-aware metrics that measure how well the system surfaces 
relevant items to users.
"""

evaluation_metric_bullets = [
    "<b>Precision@K</b> - Measures the proportion of the top-K recommended items that are actually relevant. "
    "This helps assess the immediate relevance of the recommendation list shown to a user.",

    "<b>Recall@K</b> - Measures how many of the relevant items are successfully captured within the top-K "
    "recommendations. This is useful when the goal is to maximize coverage of items a user is likely to engage with.",

    "<b>NDCG@K</b> - Normalized Discounted Cumulative Gain evaluates both relevance and ranking position, giving "
    "higher importance to relevant items that appear near the top of the recommendation list. This is particularly "
    "useful for ranked recommendation systems where ordering matters.",

    "<b>Business interpretation</b> - Together, these metrics provide a balanced view of recommendation quality by "
    "measuring top-list accuracy, coverage of relevant items, and the ranking effectiveness of the final output."
]

validation_note = """
In the implemented project, the data management workflow also demonstrates operational quality control through an 
automated validation-and-remediation stage. The validation run identified issues in the initial pass and then 
successfully revalidated the corrected data, moving from FAIL to PASS. This strengthens confidence that downstream 
EDA, feature engineering, and model training stages will consume cleaner and more reliable inputs.
"""

doc = SimpleDocTemplate(
    str(output_path),
    pagesize=A4,
    rightMargin=0.8 * inch,
    leftMargin=0.8 * inch,
    topMargin=0.8 * inch,
    bottomMargin=0.8 * inch,
)

story = [
    Paragraph(title, title_style),
    Paragraph(f"<b>Course Name:</b> {course_name}", meta_style),
    Paragraph(f"<b>Assignment Title:</b> {assignment_title}", meta_style),
    Paragraph(f"<b>Assignment:</b> {assignment_group}", meta_style),
    Spacer(1, 12),

    Paragraph("<b>Team Members</b>", heading_style),
    team_table,
    Spacer(1, 18),

    Paragraph("1. Problem Definition", heading_style),
    Paragraph(problem_definition, body_style),

    Paragraph("2. Objectives", heading_style),
    Paragraph(objectives, body_style),

    Paragraph("3. Key Data Sources and Their Attributes", heading_style),
    Paragraph(data_sources_intro, body_style),
]

for item in data_source_bullets:
    story.append(Paragraph(f"• {item}", bullet_style))

story.extend([
    Spacer(1, 8),
    Paragraph("4. Expected Outputs from the Pipeline", heading_style),
    Paragraph(expected_outputs_intro, body_style),
])

for item in expected_output_bullets:
    story.append(Paragraph(f"• {item}", bullet_style))

story.extend([
    Spacer(1, 8),
    Paragraph("5. Evaluation Metrics", heading_style),
    Paragraph(evaluation_metrics_intro, body_style),
])

for item in evaluation_metric_bullets:
    story.append(Paragraph(f"• {item}", bullet_style))

story.extend([
    Spacer(1, 8),
    Paragraph("6. Implementation Note", heading_style),
    Paragraph(validation_note, body_style),
])

doc.build(story)

print(f"PDF created successfully: {output_path}")
```
